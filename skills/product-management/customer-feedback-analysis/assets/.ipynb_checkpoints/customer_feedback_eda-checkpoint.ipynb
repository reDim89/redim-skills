# Customer feature requests — 2026 exploratory analysis

The dataset in `feature_requests_2026.csv` was built by asking Claude Code to read three
sources through their MCP servers — Granola (meeting notes), Slack (community channel) and
Pipedrive (sales notes) — and to score every feature mention on tone, customer LTV,
lifecycle stage, competitor references and churn linkage.

**This is mock data.** It is shaped like the real thing so the notebook runs end to end,
but nothing here describes a real company or customer.

Three questions:

1. Which features drive the most churn for **leads**, and which for **loyal customers**?
2. Which three features **grew** the most in mention score this year, and which three **declined**?
3. Which **competitors** come up the most, and what is the cost of churn attached to each?


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# --- chart styling -----------------------------------------------------------
# Categorical slots are assigned in fixed order and never cycled.
SERIES = ["#2a78d6", "#eb6834", "#1baf7a"]   # blue, orange, aqua
SURFACE, INK, MUTED, GRID = "#fcfcfb", "#0b0b0b", "#898781", "#e1e0d9"

plt.rcParams.update({
    "figure.facecolor": SURFACE, "axes.facecolor": SURFACE,
    "font.family": "sans-serif", "font.size": 10,
    "axes.edgecolor": GRID, "axes.labelcolor": INK, "axes.titlecolor": INK,
    "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.8,
    "xtick.color": MUTED, "ytick.color": MUTED,
    "axes.spines.top": False, "axes.spines.right": False,
    "figure.dpi": 110,
})

usd = FuncFormatter(lambda v, _: f"${v/1000:,.0f}k")

df = pd.read_csv("feature_requests_2026.csv")
df["month"] = pd.PeriodIndex(df["month"], freq="M")
print(f"{len(df)} rows · {df['feature'].nunique()} features · "
      f"{df['month'].min()} to {df['month'].max()}")
df.head()
```

## Sanity check

Before drawing anything, confirm the grain of the table: one row is *one feature, in one
month, from one lifecycle stage*. `weighted_mention_score` is already a weighted average
across that cohort's mentions, so it is averaged — never summed — when aggregating.
`mentions` and `lost_revenue_usd` are counts and money, so those do add up.

```python
print("Rows per (month, feature, stage) — should be 1:")
print(df.groupby(["month", "feature", "lifecycle_stage"]).size().value_counts(), "\n")

print("Missing values per column:")
print(df.isna().sum(), "\n")

print(f"Churn-linked rows: {int(df.churn_linked.sum())} of {len(df)}")
print(f"Total attributed lost revenue: ${df.lost_revenue_usd.sum():,}")
```

## Q1 — Which features drive churn, for leads vs loyal customers?

"Churn" means two different things depending on where the customer sits. A **lead** that
walks away is revenue never won; a **loyal customer** that leaves is revenue lost. Both are
in `lost_revenue_usd`, separated by `lifecycle_stage`, so they are read side by side rather
than pooled.

```python
churn = df[df.churn_linked == 1]

by_stage = (churn.groupby(["lifecycle_stage", "feature"])
                 .agg(lost_revenue=("lost_revenue_usd", "sum"),
                      churn_events=("churn_linked", "sum"))
                 .reset_index())

for stage in ["lead", "loyal customer"]:
    top = (by_stage[by_stage.lifecycle_stage == stage]
           .nlargest(5, "lost_revenue")
           .assign(share=lambda d: d.lost_revenue / by_stage[by_stage.lifecycle_stage == stage]
                                                     .lost_revenue.sum()))
    print(f"\n=== {stage.upper()} — top 5 by attributed lost revenue ===")
    for _, r in top.iterrows():
        n = int(r.churn_events)
        print(f"  {r.feature:<28} ${r.lost_revenue:>8,.0f}  "
              f"({r.share:5.1%} of stage total, {n} event{'s' if n != 1 else ''})")
```

```python
# Grouped bars: one bar pair per feature, leads vs loyal customers.
stages = ["lead", "loyal customer"]
pivot = (by_stage[by_stage.lifecycle_stage.isin(stages)]
         .pivot(index="feature", columns="lifecycle_stage", values="lost_revenue")
         .reindex(columns=stages)
         .fillna(0))
pivot = pivot.loc[pivot.sum(axis=1).sort_values().index]

fig, ax = plt.subplots(figsize=(9, 5.2))
y = np.arange(len(pivot))
h = 0.38                      # leaves a surface gap between the two bars

for i, stage in enumerate(stages):
    ax.barh(y + (i - 0.5) * (h + 0.02), pivot[stage], height=h,
            color=SERIES[i], label=stage, zorder=3)

ax.set_yticks(y, pivot.index)
ax.xaxis.set_major_formatter(usd)
ax.set_xlabel("Attributed lost revenue")
ax.set_title("Churn cost by feature — leads vs loyal customers", pad=14, fontsize=12)
ax.grid(axis="y", visible=False)
ax.set_axisbelow(True)
ax.legend(frameon=False, loc="lower right")
fig.tight_layout()
plt.show()
```

**Read:** the two cohorts do not churn over the same things. Leads walk away over
gate-keeping features they are evaluated against during procurement; loyal customers leave
over operational gaps they hit only after living in the product. A roadmap tuned to one
cohort will quietly ignore the other.

## Q2 — What grew and what faded this year?

Mention score is averaged across lifecycle stages per month, then fitted with a simple
linear trend. The slope is in *score points per month* — positive means the topic is
getting louder.

```python
monthly = (df.groupby(["month", "feature"])["weighted_mention_score"]
             .mean().unstack("feature").sort_index())

x = np.arange(len(monthly))

def slope(series):
    """Points-per-month trend, fitted on the months that actually have data.

    A feature can go unmentioned in a given month. Interpolating those gaps would
    invent mentions that nobody made — and a leading gap would poison the fit
    outright — so the fit simply skips them.
    """
    seen = series.notna()
    if seen.sum() < 3:
        return np.nan
    return np.polyfit(x[seen], series[seen], 1)[0]

trend = pd.DataFrame({"slope_per_month": {f: slope(monthly[f]) for f in monthly.columns}})
assert trend.slope_per_month.notna().all(), "a feature has too few months to trend"
trend["first_3mo_avg"] = monthly.head(3).mean()
trend["last_3mo_avg"] = monthly.tail(3).mean()
trend["change"] = trend.last_3mo_avg - trend.first_3mo_avg
trend = trend.sort_values("slope_per_month", ascending=False).round(2)

rising = trend.head(3)
falling = trend.tail(3).iloc[::-1]

print("=== TOP 3 RISING ===")
print(rising.to_string(), "\n")
print("=== TOP 3 DECLINING ===")
print(falling.to_string())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.4), sharey=True)
labels = [str(m) for m in monthly.index]

for ax, (title, subset) in zip(axes, [("Rising", rising.index), ("Declining", falling.index)]):
    for i, feature in enumerate(subset):
        ax.plot(x, monthly[feature], color=SERIES[i], linewidth=2,
                marker="o", markersize=5, label=feature, zorder=3)
    ax.set_title(title, fontsize=11, pad=10, loc="left")
    ax.set_xticks(x, labels, rotation=45, ha="right")
    ax.grid(axis="x", visible=False)
    ax.set_axisbelow(True)
    # Park the legend in the corner the data vacates.
    ax.legend(frameon=False, fontsize=9,
              loc="upper left" if title == "Rising" else "upper right")

axes[0].set_ylabel("Mean weighted mention score")
fig.suptitle("Feature demand trend, Jan–Aug 2026", x=0.02, ha="left", fontsize=12)
fig.tight_layout()
plt.show()
```

**Read:** the rising set is a compliance and enterprise-readiness cluster — these are
the things that block deals rather than delight users. The declining set is not proof that
demand disappeared; a topic also goes quiet after it ships, or after the loudest customer
gives up asking. Check the shipping log before reading a downward slope as "nobody cares".

## Q3 — Which competitors come up, and what do they cost?

`competitors` holds a `;`-separated list, so it is exploded to one row per named
competitor before counting. "Currently" is read as the most recent three months, shown
next to the full-year total.

```python
recent_months = sorted(df.month.unique())[-3:]

comp = (df.assign(competitor=df.competitors.fillna("").str.split(";"))
          .explode("competitor"))
comp = comp[comp.competitor.astype(str).str.strip() != ""]

summary = (comp.groupby("competitor")
               .agg(mentions=("mentions", "sum"),
                    churn_cost=("lost_revenue_usd", "sum"),
                    features=("feature", "nunique"))
               .join(comp[comp.month.isin(recent_months)]
                         .groupby("competitor")["mentions"].sum()
                         .rename("mentions_last_3mo"))
               .fillna({"mentions_last_3mo": 0})
               .astype({"mentions_last_3mo": int})
               .sort_values("mentions", ascending=False))

summary["cost_per_mention"] = (summary.churn_cost / summary.mentions).round(0)
print(summary.to_string())

print("\nFeatures each competitor is named on:")
for name in summary.index:
    feats = sorted(comp[comp.competitor == name].feature.unique())
    print(f"  {name:<10} {', '.join(feats)}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
order = summary.sort_values("mentions").index
yy = np.arange(len(order))

for ax, (col, title, fmt) in zip(axes, [
        ("mentions", "Mentions across all sources", None),
        ("churn_cost", "Attributed churn cost", usd)]):
    ax.barh(yy, summary.loc[order, col], height=0.6, color=SERIES[0], zorder=3)
    ax.set_yticks(yy, order)
    ax.set_title(title, fontsize=11, pad=10, loc="left")
    ax.grid(axis="y", visible=False)
    ax.set_axisbelow(True)
    if fmt:
        ax.xaxis.set_major_formatter(fmt)
    for i, v in enumerate(summary.loc[order, col]):
        label = f"${v/1000:,.0f}k" if fmt else f"{v:,.0f}"
        ax.text(v, yy[i], "  " + label, va="center", fontsize=9, color="#52514e")

fig.suptitle("Competitive pressure, 2026 year to date", x=0.02, ha="left", fontsize=12)
fig.tight_layout()
plt.show()
```

**Read:** mention volume and cost are not the same ranking. A competitor named often in
low-stakes conversations matters less than one named rarely but always on a deal that then
dies. `cost_per_mention` is the column worth arguing about in a roadmap review.

## What to take into the roadmap session

Not conclusions — inputs. Three things this analysis can and cannot say:

- **It can rank demand.** The scoring is consistent across sources, so relative ordering is
  meaningful even where the absolute numbers are not.
- **It cannot prove causation.** A feature named in a churn conversation is correlated with
  the loss, not necessarily the reason for it. Post-hoc reasons given in a cancellation call
  are the least reliable data in the set.
- **It is backwards-looking by construction.** Nothing here sees a market that has not
  reached your customers yet.

Take it to the planning session as a challenge to the roadmap, and let the gaps it exposes
start the argument rather than end it.
